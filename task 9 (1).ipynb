```python
categorical_features = [
    'carrier',
    'origin',
    'dest',
    'month',
    'name'
]

print("Selected Categorical Features:")
print(categorical_features)
```

```python
for col in categorical_features:

    print("\nFrequency Distribution of", col)

    print(df[col].value_counts().head(15))
```

```python
for col in categorical_features:

    print("\nPercentage Distribution of", col)

    print(
        (df[col].value_counts(normalize=True) * 100)
        .round(2).head(15)
    )
```

```python
plt.figure(figsize=(15, 10))

for i, col in enumerate(['carrier', 'origin', 'month']):
    plt.subplot(2, 2, i+1)

    sns.countplot(data=df, x=col)

    plt.title("Count Plot of " + col)
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
for col in categorical_features:

    print("\nFeature:", col)

    print("Most Common Category:",
          df[col].value_counts().idxmax())

    print("Highest Frequency:",
          df[col].value_counts().max())
```
